### 2nd way to connect to google_gen_ai

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
response = model.invoke("Write me an essay about ai")
print(response.content)
```

## Streaming and Batch

```python
for chunk in model.stream("wtite 500 word essay on yourself"):
    print(f" {chunk.text} ", end="|", flush=True)
```

```python
responses = model.batch(
    [
        "Why do you even exist??",
        "What is quantum computing?",
        "If you dont know ans then why not you simply say, i dont know??"
    ],
    config={
        'max_concurrency':5 #Limit to 5 parallel calls
    }
    )

for res in responses:
    print(res)
```
