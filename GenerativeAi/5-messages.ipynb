# Message prompt

System message: Tell behaviour <br>
Human message<br>
AI message<br>
Tool message<br>

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

message = [
    SystemMessage("You are an professor, explaining things to a layman"),
    HumanMessage("Explain random forest")
]

response = model.invoke(message)
response.content
```

```python
sys_msg = SystemMessage("You are an professor, explaining things to a layman")
message = [
    sys_msg,
    HumanMessage("Explain random forest")
]

response = model.invoke(message)
print(response.content)
```

### Message Metadata

```python
sys_msg = SystemMessage("You are an professor, explaining things to a layman")

human_msg = HumanMessage(
    content="hello!!!!!",
    name="Sana",
    id="msg_123"
)
message = [
    sys_msg,
    human_msg
]

response = model.invoke(message)
print(response.content)
```
