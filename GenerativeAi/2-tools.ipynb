```python
import os
import subprocess as sp
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

os.environ["GEMINI_API_KEY"] = os.getenv("GEMINI_API_KEY")
```

```python
model = init_chat_model(
    model = "gemini-3.1-flash-lite",
    model_provider="google_genai"
    )
```

# Tools

```python
from langchain.tools import tool

@tool
def get_date()->str:
    """Get the current local system date."""
    return sp.run(["date"])


model_with_tools = model.bind_tools([get_date]) 
```

```python
response = model_with_tools.invoke("What is the date")
print(response)
```

```python
response.tool_calls
```

### Tool execution loop

```python
# Step 1
messages = [{"role":"user","content":"What is the date?"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)

# Step 2
for tool_call in ai_msg.tool_calls:
    # Execute tool
    tool_result = get_date.invoke(tool_call)
    messages.append(tool_result)


# Step 3
final_res = model_with_tools.invoke(messages)
print(final_res.text)
```
