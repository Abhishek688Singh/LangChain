```python
from pydantic import EmailStr, AnyUrl, Field, field_validator
from typing import List, Dict , Optional, Annotated

class PatientData(BaseModel):
  name : Annotated[str, Field(
      max_length=50,
      title = "Name of Patient",
      description="skjfdnsjk",
      examples=["harry", "cherry. "]
      )]
  age : int = Field(gt=0, lt=100)

  email: EmailStr

  linkedin: AnyUrl
  # alergies: Optional[List[str]] = None
  alergies : Annotated[Optional[List[str]], Field(default=None, max_length=5)]
  contactInfo: Dict[str, str]
  married: bool = False


  @field_validator('email')
  @classmethod
  def email_validater(cla, value):
    validDomains = ["sana.in","abhi.in"]
    data = value.split('@')[-1]
    if data not in validDomains:
      raise ValueError("Not a valid domain")
    return value

  @field_validator('name')
  @classmethod
  def name_validater(cla, value):
    return value.upper()


def add_patient_data(params: PatientData):
    print(f'Adding patient data: {params.name}, age:{params.age}')
    print(201)

data = {"name": "Abhishek",
        "age": 20,
        # "email":"singh!!gmail.com", #THIS IS WRONG FORMATE
        "email": "a@a.a",
        "linkedin":"https://singhabhishek.me",
         # "alergies":["nut","trea"],
        "contactInfo":{"wife":"1234567890"},
        "married": True
        }
    # {"name": "Rahul", "age": 22},
    # {"name": "Aman", "age": 21}


new_data = PatientData(**data)

add_patient_data(new_data)
```

```python
from pydantic import EmailStr, AnyUrl, Field, model_validator
from typing import List, Dict , Optional

class PatientData(BaseModel):
  name:str = Field(max_length=50)
  age : int = Field(gt=0, lt=100)
  email: EmailStr
  linkedin: AnyUrl
  alergies: Optional[List[str]] = None
  contactInfo: Dict[str, str]
  married: bool = False

  @model_validator(mode="after")
  def validate_emergency_contact(cls, model):
    if model.age > 60 and "emergency" not in model.contactInfo:
      raise ValueError("Patient greater then 60 must have an emergency contact")
    return model

def add_patient_data(params: PatientData):
    print(f'Adding patient data: {params.name}, age:{params.age}')
    print(201)


data = {"name": "Abhishek",
        "age":90,
        # "email":"singh!!gmail.com", #THIS IS WRONG FORMATE
        "email": "a@a.a",
        "linkedin":"https://singhabhishek.me",
         # "alergies":["nut","trea"],
        "contactInfo":{"wife":"1234567890"},
        "married": True
        }
    # {"name": "Rahul", "age": 22},
    # {"name": "Aman", "age": 21}


new_data = PatientData(**data)

add_patient_data(new_data)
```

#Computed fields


```python
from pydantic import EmailStr, AnyUrl, Field, computed_field
from typing import List, Dict , Optional

class PatientData(BaseModel):
  name:str = Field(max_length=50)
  age : int = Field(gt=0, lt=100)
  email: EmailStr
  weight: float
  linkedin: AnyUrl
  alergies: Optional[List[str]] = None
  contactInfo: Dict[str, str]
  married: bool = False

  @computed_field
  @property
  def bmi(self) -> float:
    return round((self.weight //5),2)

def add_patient_data(params: PatientData):
    print(f'Adding patient data: {params.name}, age:{params.age}')
    print(201)
    print(f'bmi: {params.bmi}')

data = {"name": "Abhishek",
        "age": 20,
        # "email":"singh!!gmail.com", #THIS IS WRONG FORMATE
        "email": "a@a.a",
        "weight": 70.5,
        "linkedin":"https://singhabhishek.me",
         # "alergies":["nut","trea"],
        "contactInfo":{"wife":"1234567890"},
        "married": True
        }
    # {"name": "Rahul", "age": 22},
    # {"name": "Aman", "age": 21}


new_data = PatientData(**data)

add_patient_data(new_data)
```

#Nested model

```python
class Address(BaseModel):
  country: str
  city: str
  pin: str

class PatientData(BaseModel):
  name:str = Field(max_length=50)
  age : int = Field(gt=0, lt=100)
  email: EmailStr
  address: Address


addressData = {
               "country":"India",
               "city":"Kanpur",
               "pin":"000000"
               }

newAddressData = Address(**addressData)

patientData = {"name":"hary",
               "age":56,
               "email":"q@q.q",
               "address":newAddressData}

newpatientData = PatientData(**patientData)


print(newpatientData.name)
print(newpatientData.address.city)
```
